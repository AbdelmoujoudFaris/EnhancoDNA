# 06 - Explainability

Saliency, integrated gradients and Grad-CAM on a trained (or randomly initialised) 3D CNN.

```python
import sys; sys.path.insert(0, "../src")
import torch
from enhancoai.models.cnn3d import CNN3DModel
from enhancoai.explainability.saliency import saliency_map
from enhancoai.explainability.gradcam import grad_cam_3d

model = CNN3DModel(embedding_dim=32)
voxel = torch.rand(1, 10, 16, 16, 16)
saliency = saliency_map(model, {"voxel_grid": voxel})
cam = grad_cam_3d(model, voxel)
saliency.shape, cam.shape
```
