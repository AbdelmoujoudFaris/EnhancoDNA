# 05 - AI Training

Train the 3D CNN on the synthetic demo dataset (run scripts/generate_demo_dataset.py first).

```python
import sys; sys.path.insert(0, "../src")
from pathlib import Path
import torch
from torch.utils.data import DataLoader
sys.path.insert(0, "../scripts")
from train import VoxelDataset
from enhancoai.models.cnn3d import CNN3DModel
from enhancoai.training.trainer import Trainer

dataset = VoxelDataset("../data/processed/demo_training")
loader = DataLoader(dataset, batch_size=4, shuffle=True)
model = CNN3DModel(embedding_dim=64)
trainer = Trainer(model, device=torch.device("cpu"), experiment_dir="../experiments/notebook_demo")
trainer.fit(loader, loader, epochs=1, config_dict={"note": "notebook demo"})
```
